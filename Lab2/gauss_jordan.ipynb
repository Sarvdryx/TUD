```python
import numpy as np
```

```python
def swap_rows(A, i, j):
    temp = A[i].copy()
    A[i] = A[j]
    A[j] = temp
```

```python
def multiply_row(A, i, k):
    A[i] = A[i] * k
```

```python
def add_row_multiple(A, dest_row, src_row, k):
    A[dest_row] = A[dest_row] - k*A[src_row]
```

```python
def is_zero(val, tol=1e-9):
    return abs(val) < tol
```

```python
def Gauss_Jordan_elimination(A):
    cur_row = 0
    n, m = A.shape
    for j in range (0, m-1):
        if cur_row >= n:
            break

        max_val = 0 
        pivot_row = cur_row
        for i in range (cur_row, n):
            if abs(A[i, j]) > max_val:
                max_val = abs(A[i, j])
                pivot_row = i
        
        if is_zero(max_val):
            continue
        
        swap_rows(A, cur_row, pivot_row)
        
        multiply_row(A, cur_row, 1/A[cur_row, j])

        for k in range (cur_row+1, n):
            val = float(A[k ,j])
            add_row_multiple(A, k, cur_row, val)

        cur_row += 1

    pivot_columns = []
    for i in range(n):
        for j in range(m-1):
            if not is_zero(A[i, j]):
                pivot_columns.append(j)
                break
    
    if len(pivot_columns) < n:
        print("Matrix is singular")
        return None

    for i in range(0, n-1):
        for j in range(i+1, n):
            k = pivot_columns[j]
            val = float(A[i, k])
            add_row_multiple(A, i, j, val)

    return A
```

```python
def invert_matrix(A, Gauss_Jordan_elimination):
    rows, cols = A.shape
    I = np.eye(rows)
    
    augmented_matrix = np.hstack((A, I))
    eliminated_matrix = Gauss_Jordan_elimination(augmented_matrix)
    inv_A = eliminated_matrix[:, cols:]

    return inv_A
```
